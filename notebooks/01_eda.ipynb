# 01 — Exploratory Data Analysis

Run `python -m src.data.generate_data` first to produce `data/raw/*`. This notebook only reads and visualizes — no logic lives here that isn't also in `src/`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
import pandas as pd

from src.config import load_config

config = load_config()
machines = pd.read_csv(config.paths.machines_file)
sensors = pd.read_parquet(config.paths.sensor_file)
failures = pd.read_csv(config.paths.failures_file, parse_dates=["failure_timestamp"])
sensors.shape, machines.shape, failures.shape
```

```python
sensors[config.sensor_columns].describe()
```

```python
example_machine = sensors["machine_id"].unique()[0]
subset = sensors[sensors["machine_id"] == example_machine].set_index("timestamp")
subset[config.sensor_columns].plot(subplots=True, figsize=(10, 14), title=example_machine)
plt.tight_layout()
plt.show()
```

```python
failures["failure_type"].value_counts().plot(kind="bar", title="Failure type distribution")
plt.show()
```
